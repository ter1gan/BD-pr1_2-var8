```python
import pandas as pd
import numpy as np

data = {
    'ID': range(1, 11),
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva',
             'Frank', 'Grace', 'Helen', 'Ivan', 'Julia'],
    'Age': [23, 35, 29, 40, 28, 45, 32, 27, 38, 31],
    'City': ['Moscow', 'SPb', 'Moscow', 'Kazan', 'SPb',
             'Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb'],
    'Salary': [50000, 70000, 60000, 80000, 55000,
               90000, 65000, 58000, 75000, 62000]
}

df = pd.DataFrame(data)
print("Исходный DataFrame:")
print(df)
```

```python
print("\n1. Первые 5 строк:")
print(df.head())
```

```python
print("\n2. Строки с чётными индексами:")
print(df.iloc[::2])
```

```python
print("\n3. Строки, где длина имени > 5:")
print(df[df['Name'].str.len() > 5])
```

```python
mean_age = df['Age'].mean()
print(f"\n4. Средний возраст: {mean_age:.2f}")
```

```python
df['Senior'] = df['Age'] >= 35
print("\n5. DataFrame с новым столбцом Senior:")
print(df)
```
